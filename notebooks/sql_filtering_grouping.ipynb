# SQL Filtering, Grouping & Aggregation

Demonstration of WHERE, GROUP BY, HAVING, ORDER BY, and RANK using the actual HirePulse transaction dataset.

## 1. Load Database

```
import sqlite3
import pandas as pd
from pathlib import Path

ROOT = Path("..")
DB_PATH = ROOT / "database" / "filtering_grouping.db"
conn = sqlite3.connect(DB_PATH)
print("Connected to:", DB_PATH)
```

## 2. WHERE Filtering

WHERE filters individual rows before aggregation.

```
query = (ROOT / "queries" / "filtering_grouping" / "where_filtering.sql").read_text()
where_result = pd.read_sql_query(query, conn)
where_result
```

## 3. GROUP BY and Aggregation

GROUP BY defines the unit at which aggregate functions are calculated.

```
query = (ROOT / "queries" / "filtering_grouping" / "group_by_aggregation.sql").read_text()
group_result = pd.read_sql_query(query, conn)
group_result
```

## 4. HAVING Filtering

HAVING filters groups after aggregation.

```
query = (ROOT / "queries" / "filtering_grouping" / "having_filtering.sql").read_text()
having_result = pd.read_sql_query(query, conn)
having_result
```

## 5. WHERE + HAVING

WHERE filters rows first, while HAVING filters the resulting groups.

```
query = (ROOT / "queries" / "filtering_grouping" / "where_having_combined.sql").read_text()
combined_result = pd.read_sql_query(query, conn)
combined_result
```

## 6. ORDER BY and RANK

RANK assigns revenue position and ORDER BY surfaces the highest-revenue customers first.

```
query = (ROOT / "queries" / "filtering_grouping" / "ranking.sql").read_text()
ranking_result = pd.read_sql_query(query, conn)
ranking_result
```

## 7. Validation

```
assert (where_result["total_revenue"] > 0).all()
assert (group_result["monthly_revenue"] > 0).all()
assert (having_result["total_revenue"] > 1000).all()
assert (combined_result["total_revenue"] > 1000).all()
assert ranking_result["revenue_rank"].notna().all()
assert len(ranking_result) <= 20
print("✓ All five SQL tasks validated successfully")
```

## 8. Data Limitation

The actual transactions.csv contains only customer_id, transaction_date, and amount. It does not contain transaction_status, customer_type, or industry. Those fields were therefore not fabricated. The filtering and aggregation queries operate on the transaction dataset independently.
